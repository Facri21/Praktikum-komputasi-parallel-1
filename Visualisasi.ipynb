```python
import matplotlib.pyplot as plt

process_counts = [1, 2, 4, 8, 16,32, 64, 128, 256, 512]
speedups = [1.0383,1.9540,3.1473,4.1366,5.2607,
            4.1993,3.7740,3.0883,1.9322,0.9999]

plt.figure(figsize=(10, 6))
plt.plot(process_counts, speedups, 'bo-', label='Speedup Aktual')
plt.plot(process_counts, process_counts, 'r--', label='Speedup Ideal')

plt.xscale('log', base=2)
plt.xticks(process_counts, process_counts)

plt.xlabel('Jumlah process')
plt.ylabel('Speedup')

plt.title('Speedup vs Jumlah process')
plt.legend()
plt.grid(True)
plt.savefig('speedup_analysis.png')
plt.show()

#perbaiki kodenya ni, dimana salahnya
```

```python
import matplotlib.pyplot as plt

samples = ["100.000", "500.000", "1.000.000", "5.000.000", "10.000.0000"]
waktu_eksekusi = [0.0854, 0.0830, 0.1118, 0.2780, 0.4596]

plt.figure(figsize=(10,5))
plt.plot(samples, waktu_eksekusi, marker='o')
plt.xticks(samples)

plt.xlabel("Samples")
plt.ylabel("Waktu Eksekusi")

plt.title("Waktu Eksekusi vs Ukuran Sample")
plt.savefig('waktueksekusi_vs_sample.png')

plt.grid()

plt.show()
```

```python
import matplotlib.pyplot as plt

speed_up = ["1.0383", "1.9540", "3.1473", "4.1366", "5.2607",
           "4.1993", "3.7740", "3.0883", "1.9322", "0.9999"]
efficiency = [1.0383, 0.9770, 0.7868, 0.5170, 0.3287,
             0.1312, 0.0589, 0.0241, 0.0075, 0.0019]

plt.figure(figsize=(10,5))
plt.plot(speed_up, efficiency, marker='o')
plt.xticks(speed_up)

plt.xlabel("Speed UP")
plt.ylabel("Efficency")

plt.title("Speed Up vs Efficiency")
plt.savefig('speedup_vs_efficiency.png')

plt.grid()

plt.show()
```
